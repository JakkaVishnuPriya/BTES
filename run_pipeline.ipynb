## 4. Quick visual check (optional)
Overlay a predicted, instance-separated tree mask on the source tile for a
sanity check.

```python
import glob, os
import numpy as np
import cv2
import rasterio
import matplotlib.pyplot as plt
from matplotlib import cm

sample_png = sorted(glob.glob(os.path.join(CFG.TEST_DIR, "*.png")))[4]
name = os.path.splitext(os.path.basename(sample_png))[0]

img = cv2.cvtColor(cv2.imread(sample_png), cv2.COLOR_BGR2RGB)

with rasterio.open(os.path.join(CFG.INSTANCE_DIR, f"{name}_instances.tif")) as src:
    instances = src.read(1)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(img); ax[0].set_title("Input tile"); ax[0].axis("off")
ax[1].imshow(img); ax[1].imshow(np.ma.masked_where(instances == 0, instances), cmap=cm.get_cmap("tab20"), alpha=0.6)
ax[1].set_title(f"Separated trees (n={instances.max()})"); ax[1].axis("off")
plt.tight_layout(); plt.show()
```
